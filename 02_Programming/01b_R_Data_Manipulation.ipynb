## 1. Vectors & Vectorized Math
Vectors are the fundamental building blocks of R.


```r
# Create numeric and character vectors
gene_symbols <- c("ACTB", "GAPDH", "MALAT1", "MKI67", "CD3E")
counts <- c(1240, 3450, 8900, 45, 310)

# Vectorized arithmetic (e.g. normalize by total library size)
total_reads <- sum(counts)
cpm <- (counts / total_reads) * 1e6
names(cpm) <- gene_symbols

cat("Counts Per Million (CPM):
")
print(round(cpm, 2))

# Subsetting with boolean conditions
high_expressed <- cpm[cpm > 50000]
cat("
Highly expressed genes (>50k CPM):
")
print(high_expressed)

```

## 2. Matrices (2D Homogeneous Grids)
Gene expression matrices store counts where rows = genes and columns = samples/spots.


```r
# Construct a 4x3 expression count matrix
set.seed(42)
expr_matrix <- matrix(
  rpois(12, lambda = 50), 
  nrow = 4, 
  ncol = 3,
  dimnames = list(
    c("GeneA", "GeneB", "GeneC", "GeneD"),
    c("Sample_1", "Sample_2", "Sample_3")
  )
)

print(expr_matrix)

# Row and Column summary statistics
cat("
Gene Average Expression (rowMeans):
")
print(rowMeans(expr_matrix))

cat("
Sample Total Sequencing Depth (colSums):
")
print(colSums(expr_matrix))

```

## 3. Lists (Heterogeneous Collections)
Lists hold diverse objects together (e.g. raw counts, metadata, PCA results).


```r
# Create a multi-modal bioinformatics container
sample_obj <- list(
  sample_id = "Visium_Tissue_01",
  spot_count = 2450,
  dimensions = c(50, 50),
  qc_passed = TRUE,
  top_markers = c("EPCAM", "CD3D", "MS4A1")
)

# Accessing list elements
cat("Sample ID:", sample_obj$sample_id, "
")
cat("Top Markers:", paste(sample_obj$top_markers, collapse = ", "), "
")

```

## 4. Data Frames & Factors
Data frames are 2D tables where each column can have a different data type. Factors store categorical variables.


```r
# Sample metadata table
metadata <- data.frame(
  Sample_ID = paste0("Sample_", 1:6),
  Condition = factor(c("Control", "Control", "Control", "Treated", "Treated", "Treated")),
  Batch = factor(c("B1", "B2", "B1", "B2", "B1", "B2")),
  Total_Reads = c(2.4e6, 2.1e6, 2.8e6, 3.1e6, 2.9e6, 3.4e6),
  Mito_Percent = c(4.2, 5.1, 3.9, 12.4, 4.8, 5.0)
)

print(metadata)

# Filtering samples based on multi-column criteria
pass_qc <- subset(metadata, Mito_Percent < 10 & Total_Reads >= 2.2e6)
cat("
Samples passing QC filter:
")
print(pass_qc)

```

## 5. Merging & Handling Missing Values (`NA`)
Biological data often contains missing values that need cleaning before downstream statistics.


```r
# Handling missing values
tpm_values <- c(12.4, NA, 45.1, 0.0, NA, 88.2)

# Check NA positions
cat("Is NA vector:", is.na(tpm_values), "
")

# Compute mean excluding NAs
mean_val <- mean(tpm_values, na.rm = TRUE)
cat("Clean mean TPM:", round(mean_val, 2), "
")

# Merging metadata with sample QC scores
clinical_data <- data.frame(
  Sample_ID = c("Sample_1", "Sample_2", "Sample_3", "Sample_4"),
  Age = c(55, 62, 48, 71),
  Response = c("Responder", "Non-Responder", "Responder", "Non-Responder")
)

merged_df <- merge(metadata, clinical_data, by = "Sample_ID", all.x = TRUE)
print(merged_df)

```
