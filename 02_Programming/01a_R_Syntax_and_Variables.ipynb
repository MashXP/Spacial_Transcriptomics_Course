## 2. Atomic Data Types
R has 5 fundamental atomic data types: `numeric`, `integer`, `complex`, `character`, and `logical`.


```r
# Numeric (default floating point)
gene_tpm <- 42.5

# Integer (explicit with L)
read_count <- 120000L

# Character (text)
gene_symbol <- "GAPDH"

# Logical (Boolean TRUE / FALSE)
is_expressed <- gene_tpm > 10

# Check data types with class() and typeof()
cat("Class of gene_tpm:", class(gene_tpm), "
")
cat("Class of read_count:", class(read_count), "
")
cat("Class of gene_symbol:", class(gene_symbol), "
")
cat("Class of is_expressed:", class(is_expressed), "
")

```

## 3. String Manipulation & Pattern Matching (Regex)
Bioinformatics requires heavy string processing to format gene names, sample barcodes, and fastq headers.


```r
# 1. Combining strings with paste() and paste0()
sample_id <- "Sample_A"
replicate <- 1
full_name <- paste(sample_id, "rep", replicate, sep = "_")
print(full_name) # "Sample_A_rep_1"

# 2. String length
barcode <- "AAACCTGAGACTTTCG-1"
cat("Barcode length:", nchar(barcode), "nucleotides
")

# 3. Substring extraction (Extract 16-bp cell barcode)
cell_bc <- substr(barcode, start = 1, stop = 16)
cat("Cell Barcode:", cell_bc, "
")

# 4. Pattern matching with grep and grepl
gene_list <- c("ACTB", "GAPDH", "MT-CO1", "MT-ND1", "TP53", "MT-ATP6")

# Find mitochondrial genes (starts with MT-)
mito_indices <- grep("^MT-", gene_list)
cat("Mitochondrial gene indices:", mito_indices, "
")

mito_genes <- gene_list[mito_indices]
cat("Mitochondrial genes:", mito_genes, "
")

# 5. String replacement with sub() and gsub()
clean_names <- gsub("^MT-", "Mito_", gene_list)
print(clean_names)

```

## 4. Conditional Logic (`if`, `else if`, `else` & `ifelse`)
Conditional branches allow automated filtering and quality gating.


```r
# Quality control gate example
mito_percent <- 8.5 # % mitochondrial reads

if (mito_percent > 20) {
  status <- "FAIL: High mitochondrial contamination / dying cell"
} else if (mito_percent > 10) {
  status <- "WARNING: Moderate mitochondrial fraction"
} else {
  status <- "PASS: High quality cell"
}
cat("QC Status:", status, "
")

# Vectorized ifelse() across an entire vector
umis <- c(500, 2400, 150, 8900, 420)
qc_flags <- ifelse(umis >= 500, "Keep", "Drop")
print(data.frame(UMI_Count = umis, Decision = qc_flags))

```

## 5. Loops and Custom Functions
Functions encapsulate reproducible analysis logic.


```r
# Custom function to calculate Relative Log Expression (RLE) / Log2 Fold Change
calculate_log2fc <- function(treatment_mean, control_mean, pseudocount = 1) {
  # Defensive check
  if (treatment_mean < 0 || control_mean < 0) {
    stop("Expression values cannot be negative!")
  }
  
  log2fc <- log2((treatment_mean + pseudocount) / (control_mean + pseudocount))
  return(round(log2fc, 3))
}

# Test function
fc_val <- calculate_log2fc(treatment_mean = 145.2, control_mean = 32.1)
cat("Calculated Log2 Fold Change:", fc_val, "
")

```

## 💡 Practice Challenge: Mini Bioinformatic Exercise
**Task:** Write a function `filter_genes(genes, min_len)` that:
1. Removes any gene with name length $< min\_len$.
2. Converts all gene names to UPPERCASE using `toupper()`.
3. Returns a clean vector.



```r
# Your Solution:
filter_genes <- function(genes, min_len = 3) {
  # 1. Filter by character length
  valid_genes <- genes[nchar(genes) >= min_len]
  # 2. Uppercase
  return(toupper(valid_genes))
}

# Test data:
raw_genes <- c("a", "actb", "gapdh", "tp53", "x", "egfr")
filtered_output <- filter_genes(raw_genes, min_len = 3)
print(filtered_output) # Expected: "ACTB" "GAPDH" "TP53" "EGFR"

```
